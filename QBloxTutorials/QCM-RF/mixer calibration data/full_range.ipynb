```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data_0 = np.loadtxt(r"C:\Users\BaughLaflamme\Desktop\Qblox Master Folder\QBlox Tutorials\QCM-RF\mixer calibration data\Nov 14 2024\full_range_no_cal_before_manual.csv", delimiter=",", skiprows=45)
data_1 = np.loadtxt(r"C:\Users\BaughLaflamme\Desktop\Qblox Master Folder\QBlox Tutorials\QCM-RF\mixer calibration data\Nov 14 2024\full_range_after_manual_cal.csv", delimiter = ",", skiprows=45)
```

```python
freqs_0 = data_0[:,0]
dbm_0 = data_0[:,1]
freqs_1 = data_1[:,0]
dbm_1 = data_1[:,1]
```

```python
fig, ax = plt.subplots(1, 1, figsize = (14, 6))
ax.grid(ls="--", alpha = 0.9)
ax.plot(freqs_0, dbm_0, alpha = 0.9, label = "Before manual calibration")
ax.plot(freqs_1, dbm_1, alpha = 0.9, label = "After manual calibration")
ax.set_title("Mixer calibration of QCM-RF")
ax.set_xlabel("Frequency [GHz]")
ax.set_ylabel("Output [dBm]")
ax.set_ylim([-90,5])
plt.legend(loc = "lower left")
plt.show()
```
