```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
path0_datas = []
path1_datas = []
steps = np.arange(0, 10)

# Load data from the text files
for step in steps:
	path0_datas.append(np.loadtxt(f"step{step} path0.txt")) 
	path1_datas.append(np.loadtxt(f"step{step} path1.txt")) 
```

```python
# Plotting
fig, (ax0, ax1) = plt.subplots(nrows = 2, ncols = 1, figsize = (15, 9))
for data in path0_datas:
	ax0.plot(data)
for data in path1_datas:
	ax1.plot(data)
plt.subplots_adjust(hspace=0.3) # Add space between plots
ax0.set_title("Spin Funnel Simulation Data\n\nGate 0")
ax1.set_title("Gate 1")
ax0.set_ylabel("Input [V]")
ax1.set_ylabel("Input [V]")
ax0.set_xlabel("Time [ns]")
ax1.set_xlabel("Time [ns]")
ax0.grid(ls = '--')
ax1.grid(ls = '--')
plt.show()
```
