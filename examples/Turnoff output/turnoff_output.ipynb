I noticed when you turn off the Cluster using the switch on the back of the cluster, an output came from the QRM output, which is plotted here.

Author: Kyle MacRobbie

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import periodogram
```

```python
data = np.loadtxt(r"turnoff_output.txt", skiprows = 3)
t = np.linspace(0, 50000, len(data))

lo = 0
hi = 2000
fig, ax = plt.subplots(1, 1, figsize = (14, 4))
ax.plot(t[lo:hi], data[lo:hi])
ax.set_ylabel("Signal [V]")
ax.set_xlabel("Time [ns]")
ax.set_title("QRM Turnoff output")
ax.grid(ls = "--")
plt.show()
```

```python
# Getting a periodogram of the data
x, y = periodogram(data, 10e9)
y = np.sqrt(np.array(y))

hi = 50000
lo = 500

fig, ax = plt.subplots(1, 1, figsize = (14, 5))
ax.plot(x[lo:hi], y[lo:hi])
ax.set_xscale('log')
ax.set_yscale('log')
ax.grid(ls = '--')
ax.set_xlabel("Frequency [Hz]")
ax.set_ylabel("PSD [V/sqrt(Hz)]")
ax.set_ylim([1e-10, 1e-4])
plt.show()
```
