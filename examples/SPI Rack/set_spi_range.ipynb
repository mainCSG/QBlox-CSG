Run this notebook in order to set the range of the two D5a modules

The range can be set to one of the following:
- +/- 4 V
- +/- 2 V
- 0 V - 4 V

#### Module Addresses
These are the hardcoded addresses for the D5a modules. If you take the front panels off of the SPI rack, the address is on the module's label.

```python
D5A_ADDRESS_1 = 8 #First D5a module from the left
D5A_ADDRESS_2 = 7 #Second D5a moduel from the left
```

```python
# Connect to the instrument
from qblox_instruments import SpiRack
spi = SpiRack("SpiRack", "COM3")
print(spi.IDN())
spi.add_spi_module(D5A_ADDRESS_1, "D5a", "module1")
spi.add_spi_module(D5A_ADDRESS_2, "D5a", "module2")
```

```python
# Changing the range of the output
# You can manually select +/- 4V (range_4V_bi), +/- 2V (range_2V_bi) or 0 to +4V (range_4V_uni)
spi.module1.dac0.span('range_4V_bi')
print(f"Module 1 range: {spi.module1.dac0.span()}")
spi.module2.dac0.span('range_4V_bi')
print(f"Module 2 range: {spi.module2.dac0.span()}")
```

You can also change the voltage levels maunally like below:

```python
spi.module2.dac8.voltage(0.05)
```

```python
# closing connection to the instrument
# Whatever values the voltages are set to will remain the same after the close command is run, so i would suggest setting it to 
# 0 if you have set it to anything else
spi.close()
```
