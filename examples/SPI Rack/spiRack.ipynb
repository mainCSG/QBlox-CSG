#### Module Addresses
These are the hardcoded addresses for the D5a modules. If you take the front panels off of the SPI rack, the address is on the module's label.

```python
D5A_ADDRESS_1 = 8 #First D5a module from the left
D5A_ADDRESS_2 = 7 #Second D5a moduel from the left
```

```python
from qblox_instruments import SpiRack
import numpy as np
import time

spi = SpiRack("SpiRack", "COM3")
print(spi.IDN())

#Parameters
#1 Module address
#2 Module type
#3 Module name used to access the module
spi.add_spi_module(D5A_ADDRESS_1, "D5a", "module1")
spi.add_spi_module(D5A_ADDRESS_2, "D5a", "module2")
```

```python
def sweep(start_p, end_p, step_size, dac_sel):

	dac1_init_voltage = int(spi.module1.dac1.voltage())
	dac2_init_voltage = int(spi.module1.dac2.voltage())

	spi.module1.dac1.ramp_rate(2)
	spi.module1.dac1.ramp_max_step(step_size)
	spi.module1.dac1.ramping_enabled(True)

	spi.module1.dac2.ramp_rate(2)
	spi.module1.dac2.ramp_max_step(step_size)
	spi.module1.dac2.ramping_enabled(True)

	if (start_p != dac1_init_voltage) and (dac_sel == 1): 
		print("Setting initial voltage on DAC 1")
		spi.module1.dac1.voltage(start_p)
		while spi.module1.dac1.is_ramping():
			time.sleep(0.01)
			
	if (start_p !=  dac2_init_voltage) and (dac_sel == 2):
		print("Setting initial voltage on DAC 2")
		spi.module1.dac2.voltage(start_p)
		while spi.module1.dac2.is_ramping():
			time.sleep(0.01)
				
	if dac_sel == 1:
		spi.module1.dac1.voltage(end_p)
		while spi.module1.dac1.is_ramping():
			time.sleep(0.01)
		print("Setting final voltage on DAC 1")
	if dac_sel == 2:
		spi.module1.dac2.voltage(end_p)
		while spi.module1.dac2.is_ramping():
			time.sleep(0.01)

		print("Setting final voltage on DAC 2")
```

```python
sweep(1, 0, 0.1, 1)
```

```python
# Changing the range of the output
# You can manually select +/- 4V (range_4V_bi), +/- 2V (range_2V_bi) or 0 to +4V (range_4V_uni)
#spi.module1.dac2.span('range_4V_bi')
#spi.module1.dac2.span()
sweep(0, 1, 0.001, 1)
sweep(0, 1, 0.001, 2)
```
