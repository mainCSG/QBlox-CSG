# SPI Rack
Experiment to test connections to the SPI rack. The outputs of each module were connected to an oscilloscope for viewing.

Qblox SPI rack setup documentation: https://qblox-qblox-instruments.readthedocs-hosted.com/en/main/getting_started/setup.html#installation

#### Module Addresses
These are the hardcoded addresses for the D5a modules. If you take the front panels off of the SPI rack, the address is on the module's label.

```python
D5A_ADDRESS_1 = 8 #First D5a module from the left
D5A_ADDRESS_2 = 7 #Second D5a moduel from the left
```

```python
from qblox_instruments import SpiRack

spi = SpiRack("SpiRack", "COM3")
print(spi.IDN())

#Parameters
#1 Module address
#2 Module type
#3 Module name used to access the module
spi.add_spi_module(D5A_ADDRESS_1, "D5a", "module1")
spi.add_spi_module(D5A_ADDRESS_2, "D5a", "module2")
```

```python
#To set a specific output channel:
#spi.module1.dacX.voltage(1.0)
#Where X is the channel index 

#Channels are physically labeled starting at 1,
#but referencing them in software is 0 based indexing

#Sets the parameter value
spi.module1.dac0.voltage(0.1)
#Prints the parameter value
spi.module1.dac0.voltage()

spi.module2.dac0.voltage(0.1)
spi.module2.dac0.voltage()
```
