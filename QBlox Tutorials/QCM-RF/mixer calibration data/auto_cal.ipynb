```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
day_0 = np.loadtxt(r"C:\Users\BaughLaflamme\Desktop\Qblox Master Folder\QBlox Tutorials\QCM-RF\mixer calibration data\Nov 14 2024\auto_cal_before_cal.csv", delimiter=",", skiprows=45)
day_1 = np.loadtxt(r"C:\Users\BaughLaflamme\Desktop\Qblox Master Folder\QBlox Tutorials\QCM-RF\mixer calibration data\Nov 14 2024\auto_cal_after_both.csv", delimiter = ",", skiprows=45)
day_2 = np.loadtxt(r"C:\Users\BaughLaflamme\Desktop\Qblox Master Folder\QBlox Tutorials\QCM-RF\mixer calibration data\Nov 14 2024\auto_cal_after_sb_again.csv", delimiter = ",", skiprows=45)
```

```python
hi = 858
lo = 180

freqs_0 = day_0[:,0][lo:hi]
dbm_0 = day_0[:,1][lo:hi]
freqs_1 = day_1[:,0][lo:hi]
dbm_1 = day_1[:,1][lo:hi]
freqs_2 = day_2[:,0][lo:hi]
dbm_2 = day_2[:,1][lo:hi]
```

```python
fig, ax = plt.subplots(1, 1, figsize = (14, 6))
ax.grid(ls="--", alpha = 0.9)
ax.plot(freqs_0, dbm_0, alpha = 1.0, label = "Before any calibration")
ax.plot(freqs_1, dbm_1, alpha = 0.9, label = "After LO and SB calibration")
ax.plot(freqs_2, dbm_2, alpha = 0.7, label = "After SB calibration again")
ax.set_title("Mixer calibration of QCM-RF")
ax.set_xlabel("Frequency [Hz]")
ax.set_ylabel("Output [dBm]")
ax.set_ylim([-95,5])
plt.legend(loc = "lower left")
plt.show()
```

```python
fig, ax = plt.subplots(1, 1, figsize = (14, 6))
ax.grid(ls="--", alpha = 0.9)
ax.plot(freqs_0, dbm_0, alpha = 0.9, label = "Before any calibration")
ax.plot(freqs_2, dbm_2, alpha = 0.9, label = "After auto calibration")
ax.set_title("Mixer calibration of QCM-RF")
ax.set_xlabel("Frequency [GHz]")
ax.set_ylabel("Output [dBm]")
ax.set_ylim([-95,5])
plt.legend(loc = "lower left")
plt.show()
```
