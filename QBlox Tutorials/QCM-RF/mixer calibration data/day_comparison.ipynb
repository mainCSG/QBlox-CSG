```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
day_0 = np.loadtxt(r"C:\\Users\\BaughLaflamme\\Desktop\\Qblox Master Folder\\QBlox Tutorials\\QCM-RF\\mixer calibration data\\Nov 12 -13 2024\\after cal fast.csv", delimiter=",")
day_1 = np.loadtxt(r"C:\\Users\\BaughLaflamme\\Desktop\\Qblox Master Folder\\QBlox Tutorials\\QCM-RF\\mixer calibration data\\Nov 12 -13 2024\\day after.csv", delimiter = ",")
day_2 = np.loadtxt(r"C:\\Users\\BaughLaflamme\\Desktop\\Qblox Master Folder\\QBlox Tutorials\\QCM-RF\\mixer calibration data\\Nov 14 2024\\two days after.csv", delimiter = ",")
day_3 = np.loadtxt(r"C:\\Users\\BaughLaflamme\\Desktop\\Qblox Master Folder\\QBlox Tutorials\\QCM-RF\\mixer calibration data\\Nov 15\\3 days after.csv", delimiter = ",", skiprows=45)
```

```python
hi = 858
lo = 180

freqs_0 = day_0[:,0][lo:hi]
dbm_0 = day_0[:,1][lo:hi]
freqs_1 = day_1[:,0][lo:hi]
dbm_1 = day_1[:,1][lo:hi]
freqs_2 = day_2[:,0][lo:hi]
dbm_2 = day_2[:,1][lo:hi]
freqs_3 = day_3[:,0][352:625]
dbm_3 = day_3[:,1][352:625]
```

```python
fig, ax = plt.subplots(1, 1, figsize = (14, 6))
ax.grid(ls="--", alpha = 0.9)
ax.plot(freqs_0, dbm_0, alpha = 1.0, label = "Day 0")
ax.plot(freqs_1, dbm_1, alpha = 0.9, label = "Day 1")
ax.plot(freqs_2, dbm_2, alpha = 0.7, label = "Day 2")
ax.plot(freqs_3, dbm_3, alpha = 0.7, label = "Day 3", color = 'r')
ax.set_title("Mixer calibration of QCM-RF: day-to-day drift comparison")
ax.set_xlabel("Frequency [GHz]")
ax.set_ylabel("Output [dBm]")
ax.set_ylim([-90,5])
plt.legend(loc = "upper left")
plt.show()
```
